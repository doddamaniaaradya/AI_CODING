Part 2: The Countries API 🌍

CELL 4 · Facts about any country

```python
import requests

country_input = input("Type a country: ").strip().lower()

# Fetch the entire database from the stable mirror endpoint
url = "https://studies.cs.helsinki.fi/restcountries/api/all"

try:
    response = requests.get(url)
    if response.status_code == 200:
        countries_list = response.json()

        # Look for the country by comparing common or official names
        match = None
        for country in countries_list:
            common_name = country.get("name", {}).get("common", "").lower()
            official_name = country.get("name", {}).get("official", "").lower()

            if country_input == common_name or country_input == official_name:
                match = country
                break

        # If found, display the attributes cleanly
        if match:
            # Mirror provides capitals inside an array, handle if empty
            capitals = match.get("capital", [])
            capital_display = ", ".join(capitals) if capitals else "N/A"

            print("\n" + match.get("flag", ""), match.get("name", {}).get("common", "Unknown"))
            print("Capital:", capital_display)
            print("People:", match.get("population", "N/A"))
            print("Part of the world:", match.get("region", "N/A"))
        else:
            print("\nHmm, the website doesn't recognize that country name!")

    else:
        print(f"Server returned an error status code: {response.status_code}")

except requests.exceptions.RequestException as e:
    print("\nReal connection error! Please verify your internet connection setup.")

```

Part 3: The Trivia API 🧠

CELL 6 · GAME: True or False Trivia!

```python
import html

data = requests.get("https://opentdb.com/api.php?amount=5&type=boolean&difficulty=easy").json()
score = 0

for q in data["results"]:
  question = html.unescape(q["question"])
  guess = input(question + " (True/False) ") # turns codes like &#039; back into '
  if guess.lower() == q["correct_answer"].lower():
    print(" ✅ Correct!")
    score = score + 1

  else:
    print(" ❌ Nope! The answer was", q["correct_answer"])
    print("Your score:", score, "out of 5")
```
