```python
import pandas as pd

pets = pd.DataFrame({
    "name": ["Bruno", "Mittens", "Nemo", "Coco", "Hopper", "Shelly"],
    "animal": ["Dog", "Cat", "Fish", "Parrot", "Rabbit", "Turtle"],
    "legs": [4, 4, 0, 2, 4, 4],
    "cuteness": [9, 10, 6, 8, 10, 7],
    "sound": ["Woof", "Meow", "Blub", "Hello!", "Squeak", "Hmm"]
})
pets
```

Pet Detective!

```python
pet = pets.sample(1).iloc[0]   # the computer secretly picks one pet
print("Clue 1: I have", pet["legs"], "legs.")
print("Clue 2: My cuteness score is", pet["cuteness"])
print("Clue 3: I say", pet["sound"])
guess = input("Which animal am I? ")
if guess.lower() == pet["animal"].lower():
  print("Correct! You are a Pet Detective!")
else: print("Oops! I was", pet["name"], "the", pet["animal"])
```

Advanced Pet Detective!

```python
score = 0

for turn in range(3):            # play 3 rounds
 pet = pets.sample(1).iloc[0]
 print("Clues:", pet["legs"], "legs,", "says", pet["sound"])
 guess = input("Which animal am I? ")
 if guess.lower() == pet["animal"].lower():
  print("Correct!")
  score = score + 1
 else: print("It was the", pet["animal"])
print("Your detective score:",score, "out of 3")
```
