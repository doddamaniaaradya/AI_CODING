```python
import pandas as pd
import matplotlib.pyplot as plt
pets = pd.DataFrame({
"name":     ["Bruno", "Mittens", "Nemo", "Coco", "Hopper", "Shelly"],
"animal":      ["Dog", "Cat", "Fish", "Parrot", "Rabbit", "Turtle"],
"legs":[4, 4, 0, 2, 4, 4],
"cuteness": [9, 10, 6, 8, 10, 7],
})
print("Pets loaded:", len(pets))
```

```python
print(pets)
```

```python
plt.bar(pets["name"], pets["cuteness"])
plt.show()
```

```python
colors = ["gold", "silver", "skyblue", "lavender", "orchid", "yellow"]
plt.bar(pets["name"], pets["cuteness"], color=colors)
plt.title("The Colorful Cuteness Chart")
plt.xlabel("Pet")
plt.ylabel("Cuteness (out of 10)")
plt.show()
```

```python
ranked = pets.sort_values("cuteness")
plt.barh(ranked["name"], ranked["cuteness"], color="gold")
plt.title("Cuteness Leaderboard")
plt.show()
```

```python
leg_counts = pets["legs"].value_counts()
print(leg_counts)
plt.pie(leg_counts, labels=leg_counts.index,
colors=["darkblue", "green", "lightblue"], autopct="%.0f%%")
plt.title("How many legs? (4, 2 or 0)")
plt.show()
```

```python
animals = ["Puppy", "Kitten", "Panda"]
scores = []
for animal in animals:
 score = int(input(f"How cute is the {animal}? (1 to 10) "))
 scores.append(score)
# add the score to the list
plt.bar(animals, scores, color=["Pink", "Yellow", "Purple"])
plt.title("My Cuteness Contest")
plt.ylim(0, 10)
plt.show()
winner = animals[scores.index(max(scores))]
print("And the winner is...", winner, "!")
```
