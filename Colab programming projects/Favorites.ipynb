```python
import random
choice = ["rock","paper","scissors"]
computer_choice = random.choice(choice)
user_choice = str(input("Write rock, paper, or scissors: "))
if computer_choice == user_choice:
  print("It's a tie")
elif computer_choice == "scissors" and user_choice == "rock":
  print("User wins")
elif computer_choice == "rock" and user_choice == "paper":
  print("User wins")
elif computer_choice == "paper" and user_choice == "scissors":
  print("User wins")
elif computer_choice == "rock" and user_choice == "scissors":
  print("Computer wins")
elif computer_choice == "paper" and user_choice == "rock":
  print("Computer wins")
elif computer_choice == "scissors" and user_choice == "paper":
  print("Computer wins")
else:
  print("Please type again")
```

```python
print("Welcome to my Pattern Store!")
option = str(input("Pick your pattern: Inverted Pyramid, Diamond, Inverted, Upright: "))
if option == "Inverted Pyramid":
  print("Give me 5000 won")
  base = 5
  for i in range(base, 0, -1):
   print(" "*(base-i), "*"* (2*i-1))
elif option == "Diamond":
  print("Give me 5000 won")
  base = 5
  for i in range(0, base, 1):
   print(" "*(base-i), "*"* (2*i-1))
  for i in range(base, 0, -1):
   print(" "*(base-i), "*"* (2*i-1))
elif option == "Inverted":
  print("Give me 5000 won")
  for i in range(10,0,-1):
   print("$"*i)
elif option ==  "Upright":
  print("Give me 5000 won")
  for i in range(0,10):
   print("$"*i)
else:
  print("That is not on the menu")
print("I hope you enjoyed my Store. Bye bye!")
```
