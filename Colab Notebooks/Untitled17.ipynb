```python
fruit=str(input("Tell me your favorite fruit: "))
for i in fruit:
  if i == 'S' or i == 's':
    print("There is an S!!!!")
    break
else:
 print("No S.")
```

```python
name=str(input("Tell me your name: "))
for name in name:
  if name == 'A':
    print("There is an A!!!!")
    break
else:
 print("No A.")
```
