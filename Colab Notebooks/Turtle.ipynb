```python
import ColabTurtle.Turtle as t
import random

# 1. Initialize the canvas (Required in Colab before drawing)
t.initializeTurtle()
t.bgcolor("black")
t.speed(10)      # Speed ranges 1 to 10
t.pensize(2)

# Cardinal angles: 0=East, 90=South, 180=West, 270=North (Colab Turtle orientation)
angles = [0, 90, 180, 270]

for _ in range(150):
    # Pick a random RGB color tuple
    r = random.randint(50, 255)
    g = random.randint(50, 255)
    b = random.randint(50, 255)
    t.color((r, g, b))

    # Set direction and step forward
    t.setheading(random.choice(angles))
    t.forward(20)
```
