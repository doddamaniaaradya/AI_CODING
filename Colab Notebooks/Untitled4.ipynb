```python
calendar=input("Write a letter choices are a, f, m, j, s, o, n, and d: ")
if calendar == "a":
 print("The months that start with A are April and August")
elif calendar == "f":
 print("The solo February")
elif calendar == "m":
  print("Partners March and May")
elif calendar == "j":
  print("Hot and cold, January, June, and July")
elif calendar == "s":
  print("Fall down September")
elif calendar == "o":
  print("Cold October")
elif calendar == "n":
  print("Coldest November")
elif calendar == "d":
  print("Snowing December")
else:
  print("Please type a letter")
```
