```python
num1=int(input("Enter your first number: "))
num2=int(input("Enter your second number: "))
op=str(input("Do you want to Add, Sub, Mul, Div, IntDiv, Mod, or FloatDiv?: "))
if op == 'Add':
  num3=num2+num1
  print(num3)
elif op == 'Sub':
  num3=num2-num1
  print(num3)
elif op == 'Mul':
  num3=num2*num1
  print(num3)
elif op == 'Div': #Integer division
  num3=num2//num1
  print(num3)
elif op == 'IntDiv':
  num3=num2//num1
  print(num3)
elif op == 'Mod':
  num3=num2%num1
  print(num3)
elif op == 'FloatDiv':
  num3=num2/num1
  print(num3)
else:
  print("Error! Please try again.")
```
